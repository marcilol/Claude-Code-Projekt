```
import os

csv_file_path = 'data/input/russell_constituents.csv'

# Read all lines from the file
with open(csv_file_path, 'r') as f:
    lines = f.readlines()

# Find the line containing the correct header and remove merge conflict markers
# The correct header appears to be 'Ticker;Name;Sector;Asset Class;Location;Exchange'
# and is usually found after the '<<<<<<< HEAD' marker and before '=======' or '>>>>>>>'

cleaned_lines = []
skip_until_header = False
for line in lines:
    if '<<<<<<< HEAD' in line:
        skip_until_header = True
        continue
    if '>>>>>>>' in line or '=======' in line:
        skip_until_header = False
        continue
    if not skip_until_header:
        cleaned_lines.append(line)

# Ensure the header is correct. Based on the output, the first valid line is the header.
# If the first line is still problematic, or if there's no header yet, ensure it's added.
if not cleaned_lines or 'Ticker;Name;Sector;Asset Class;Location;Exchange' not in cleaned_lines[0]:
    # Fallback to a known good header if the cleaning logic somehow fails or isn't perfect
    cleaned_lines.insert(0, 'Ticker;Name;Sector;Asset Class;Location;Exchange\n')

# Write the cleaned content back to the file
with open(csv_file_path, 'w') as f:
    f.writelines(cleaned_lines)

print(f"Cleaned Git merge conflict markers from {csv_file_path}.")

# Verify the content again (first 5 lines)
print(f"\nContent of {csv_file_path} after cleaning (first 5 lines):\n")
with open(csv_file_path, 'r') as f:
    for i, line in enumerate(f):
        print(line.strip())
        if i >= 4:
            break
```

```
import os

file_path = 'scripts/marci/fetch_data.py'

with open(file_path, 'r') as f:
    lines = f.readlines()

# The exact content of the line we want to insert before, to extract its indentation
target_line_content = "tickers = russell_df['Ticker'].tolist()"
# The debug line content to insert
debug_print_statement = "print('DEBUG: Columns in russell_df:', russell_df.columns)"

new_lines = []
insertion_made = False

for i, line in enumerate(lines):
    # Check if this is the problematic debug line we inserted before and remove it
    if debug_print_statement in line:
        continue # Skip adding the old debug line

    # If we found the target line and haven't inserted yet
    if target_line_content in line and not insertion_made:
        # Extract the indentation from the target line
        indentation = line.split(target_line_content)[0]
        # Insert the debug statement with the extracted indentation
        new_lines.append(f"{indentation}{debug_print_statement}\n")
        insertion_made = True

    new_lines.append(line)

# If for some reason the target line wasn't found (shouldn't happen given previous context),
# we still want to inform the user.
if not insertion_made:
    print(f"Warning: Could not find the target line '{target_line_content}' in {file_path}. Manual inspection might be needed.")

# Write the modified content back to the file
with open(file_path, 'w') as f:
    f.writelines(new_lines)

print(f"Corrected indentation and re-inserted debug print statement in {file_path}.")

# Now, re-run the script to see the debug output
!python scripts/marci/fetch_data.py
```
